```python
from helpers import *
from regression_models import *
from statistic_models import *
from lstm_models import *
from transformer_models import *

df = load_dataframe('data/2023_PM10_1g.xlsx')
```

```python
col_null_counts = df.iloc[:, 1:].isnull().sum()
least_missing_cols = col_null_counts.nlargest(10).index.tolist()
print("Columns with least missing values:", least_missing_cols)
df_subset = df[least_missing_cols]
df_subset.head()
count_nulls(df_subset)
```

```python
results = []
for col in df_subset.columns:
    metrics = evaluate_imputation(df_subset[col], impute_bfill, mask_fraction=0.03, random_state=42)
    metrics['station'] = col
    results.append(metrics)

results_df = pd.DataFrame(results).set_index('station')
print(results_df)
mean_per_error = results_df.mean(axis=0)
print(mean_per_error)
```
